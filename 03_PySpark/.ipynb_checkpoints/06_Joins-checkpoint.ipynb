```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
        .appName ("GoToSpark-Test") \
        .getOrCreate()

print(spark.version)
```

```python
df_loc = spark.read.csv("Data/nyclocations.csv",
               header=True,
               inferSchema=True)
```

```python
df_taxi=spark.read.csv("Data/nyctaxi.csv",
               header=True,
               inferSchema=True)
```

```python
df_taxi.count()
```

```python
df_loc.count()
```

```python
# to join two tables - simple join 
# common column - PU location id

df_join = df_taxi.join(df_loc) # not mentioning the common column
df_join.show(5)
```

```python
# Here the join is cross join. PU location ID is not same. it is taken as 2 columns
```

```python
df_join.count()
```

```python
# this is not the result that we need
```

```python
df_join.explain(True)
```

```python
# BroadcastNestedLoopJoin ~ cross join
```

```python
# inner join
df_join2 = df_taxi.join(df_loc, df_taxi.PUlocationID == df_loc.PUlocationID)
df_join2.show()
```

```python
df_join2.explain(True)
```

```python
# == Optimized Logical Plan ==
# Join Inner, (PUlocationID#40 = PUlocationID#17)

# BroadcastHashJoin 
```

```python
df_join2.count()
```

```python
# inner join 
df_join3 = df_taxi.join(df_loc, df_taxi.PUlocationID == df_loc.PUlocationID, "inner")
df_join3.show()
```

```python
# left join 
df_join4 = df_taxi.join(df_loc, df_taxi.PUlocationID == df_loc.PUlocationID, "left")
df_join4.count()
```

```python
# left join 
df_join4 = df_loc.join(df_taxi, df_taxi.PUlocationID == df_loc.PUlocationID, "left")
df_join4.count()
```

```python
# there is very little difference in the count even if we changed the left table. 
```

```python
df_leftanti = df_loc.join(df_taxi, df_taxi.PUlocationID == df_loc.PUlocationID, "leftanti")
df_leftanti.count()
```

### 1. Broadcast

```python
from pyspark.sql.functions import broadcast
```

```python
df_join5 = df_taxi.join(broadcast(df_loc), df_taxi.PUlocationID == df_loc.PUlocationID, "inner")
df_join5.explain()
```

```python
spark.conf.get("spark.sql.autoBroadcastJoinThreshold")
```

```python
10485760/1024/1024
```

```python
# 10mb is the default memory from broadcast
```

```python
# we can set the threshold
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", 100*1024*1024)
```

```python
spark.conf.get("spark.sql.autoBroadcastJoinThreshold")
```

```python
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1) # disable the broadcast
```

```python
104857600/1024/1024
```
