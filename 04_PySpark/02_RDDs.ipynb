```python
list1 = [1,2,3,4,5]
print(list1)
```

```python
type(list1)
```

```python
# RDDs - Resilient Distributed Dataases
```

```python
sparkcon = spark.sparkContext # we can save it in a variable, itstead of using the whole thing everytime
```

```python
rdd1 = sparkcon.parallelize(list1) 
```

```python
# Parallelize 
# used to create an rdd.
# considers the python collections
# divides the rdd into multiple paritions
```

```python
rdd1.collect() # collect is an action. only when action is mentioned it will be executed.
```

```python
rdd1.getNumPartitions() # gives the number of partitions
```

```python
# How the 5 elements in the list are allocated in the 8 partitions?
```

```python
rdd1.glom().collect()
```

```python
# We can change the number of partitions - Coalesce() to reduce & no shuffle ; and repartition() to increae/decrease and the elements shuffle
```

```python
rdd2 = rdd1.coalesce(2) # reducing the number of partitions
```

```python
rdd2.getNumPartitions()
```

```python
rdd2.glom().collect()
```

```python
rdd3=rdd1.repartition(6)
```

```python
rdd3.getNumPartitions()
```

```python
rdd3.glom().collect()
```
