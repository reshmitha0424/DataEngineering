```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
        .appName ("GoToSpark-Test") \
        .getOrCreate()

print(spark.version)
```

There is no schema in RDD. 
Converting RDD to Dataframe becasue dataframes have better performance than RDD

```python
data = [
    (1, "John", 1000000),
    (2, "Adam", 200000), 
    (3, "Sam", 300000)]
```

```python
sparkcon=spark.sparkContext
```

```python
rdd=sparkcon.parallelize(data)
```

```python
rdd.collect()
```

### Converting RDD to dataframe

```python
df = rdd.toDF(
    # schema
    ["id", "name", "salary"])
```

```python
display(df)
```

```python
df2 = spark.createDataFrame(data, ["id", "name", "salary"])
```

```python
display(df2)
```

### To explicitly define the schema

```python
from pyspark.sql.types import (StructType, StructField, IntegerType, StringType)

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True), 
    StructField("salary", IntegerType(), True)
])
```

```python

df3 = spark.createDataFrame(rdd, schema)
```

```python
display(df3)
```

```python
# print schema
df3.printSchema()
```
