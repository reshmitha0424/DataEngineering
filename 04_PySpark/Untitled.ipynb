```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
        .appName ("GoToSpark-Test") \
        .getOrCreate()

print(spark.version)
```

```python
df = spark.read.csv("Data/superstore.csv",
               header=True,
               inferSchema=True)
```

```python
from pyspark.storagelevel import StorageLevel
from pyspark.sql import functions as F
import time
```

```python
df_test = df.select("Category", "Region", "Sales")
```

```python
# running without cache
start = time.time()
df_test.count()
print("Run without cache:", time.time() -start, "seconds")
```

```python
start = time.time()
df_test.show(5)
print("Run without cache:", time.time() -start, "seconds")
```

```python
df_test.cache()
```

```python
# running with cache
start = time.time()
df_test.count()
print("Run with cache:", time.time() -start, "seconds")
```

```python
start = time.time()
df_test.show(5)
print("Run with cache:", time.time() -start, "seconds")
```

```python
# clear the cache
df_test.unpersist()
```

```python
df_test.persist(StorageLevel.MEMORY_ONLY)
```

```python
# running with cache
start = time.time()
df_test.count()
print("Run with cache:", time.time() -start, "seconds")
```

```python
df_test.persist(StorageLevel.DISK_ONLY)
```

```python
# running with cache
start = time.time()
df_test.count()
print("Run with cache:", time.time() -start, "seconds")
```

```python
df_test.persist(StorageLevel.MEMORY_AND_DISK)
```

```python
# running with cache
start = time.time()
df_test.count()
print("Run with cache:", time.time() -start, "seconds")
```
