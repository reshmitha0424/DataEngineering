### Define the Schema Explicitly

```python
from pyspark.sql.types import (StructType, StructField, IntegerType, StringType)

schema_a = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("country", StringType(), True),
    StructField("department", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("experience_years", IntegerType(), True),
    StructField("email", StringType(), True),
    StructField("status", StringType(), True),
    StructField("join_date", StringType(), True)])
    
```

```python
df = spark.read.csv("Data/Employees.csv",
               header=True,
               schema=schema_a)
```

```python
display(df)
```

### Selecting Specific columns

```python
df = spark.read.csv("Data/Employees.csv",
               header=True,
               inferSchema=True)
df.show()
```

```python
df1 = df.select("No", "First Name", "Last Name", "Gender")
```

```python
df1.show()
```

### Renaming the column names

```python
from pyspark.sql.functions import expr
```

```python
df2 = df.selectExpr(
    "No as Customer_id",
    "`First Name` as Empl_FirstName",
    "`Last Name` as Empl_LastName",
    "Gender"
)
```

```python
display(df2)
```

```python
df3 = df.select(
    expr("No as Customer_id"),
    expr("`First Name` as Empl_FirstName"),
    expr("`Last Name` as Empl_LastName"),
    "Gender")
```

```python
display(df3)
```

```python
df2.show()
```

```python
df3.show()
```
