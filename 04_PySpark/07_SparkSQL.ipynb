```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
        .appName ("GoToSpark-Test") \
        .getOrCreate()

print(spark.version)
```

```python
df = spark.read.csv("Data/superstore.csv",
               header=True,
               inferSchema=True)
```

```python
df.createOrReplaceTempView("superstore")
```

```python
sql_result = spark.sql(""" Select * from superstore """)
```

```python
sql_result.show(5)
```

```python
spark.sql(""" select `Customer ID`, Category, Sales from superstore """).show(5)
```

```python
df.select(
    "Customer ID",
    "Category",
    "Sales").show(5)
```

```python
spark.sql(""" select distinct Category from superstore """).show(5)
```

```python
df.select("Category").distinct().show()
```
