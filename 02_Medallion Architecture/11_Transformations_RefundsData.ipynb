Refunds data is in different catalog<br>
we need to create a schema in that catalog for all layers

we can create a table and save it in our bronze layer 

```sql
select * from azuresqldatabase_ecomm_catalog.dbo.refunds
```

```sql
Create table ecomm.bronze.refunds
as 
select * from azuresqldatabase_ecomm_catalog.dbo.refunds
```

```sql
Select * from ecomm.bronze.refunds
```

```sql
Select refund_id,
payment_id,
refund_timestamp,
refund_amount,
Split(refund_reason,':')[0] as refund_reason,
Split(refund_reason,':')[1] as refund_by
from ecomm.bronze.refunds
```

```sql
Select refund_id,
payment_id,
cast(date_format(refund_timestamp, 'yyyy-MM-dd') as Date) as refund_date,
date_format(refund_timestamp, 'HH-mm-ss') as refund_time,
refund_amount,
Split(refund_reason,':')[0] as refund_reason,
Split(refund_reason,':')[1] as refund_by
from ecomm.bronze.refunds
```

```sql
CREATE TABLE ecomm.silver.refunds AS
Select refund_id,
payment_id,
cast(date_format(refund_timestamp, 'yyyy-MM-dd') as Date) as refund_date,
date_format(refund_timestamp, 'HH-mm-ss') as refund_time,
refund_amount,
Split(refund_reason,':')[0] as refund_reason,
Split(refund_reason,':')[1] as refund_by
from ecomm.bronze.refunds
```
